```python
#9. Factorial using Function

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
```

```python
#10. Prime Number Checker
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
#15.Word Frequency Counter
sentence = input("Enter sentence: ")
words = sentence.split()
freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```

```python
#17.Student Result Analyzer
marks = [45, 78, 90, 35, 60]

def average(marks):
    return sum(marks) / len(marks)

def passed_students(marks):
    for mark in marks:
        if mark > 40:
            print(mark)

print("Average:", average(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))

print("Passed Students:")
passed_students(marks)
```

```python
#18.Password Strength Checker
def check_password(password):

    if (len(password) >= 8 and
        any(char.isdigit() for char in password) and
        any(char.isupper() for char in password) and
        any(char.islower() for char in password)):
        return "Strong Password"

    return "Weak Password"

pwd = input("Enter password: ")
print(check_password(pwd))
```

```python
#20.Mini Banking System
balance = 1000

def deposit(amount):
    global balance
    balance += amount
def withdraw(amount):
    global balance
    balance -= amount
def check_balance():
    print("Balance:", balance)

deposit(500)
withdraw(200)
check_balance()
```

```python
#21.HackerRank Style Problem
numbers = [10, 20, 30, 40, 50]

max_num = numbers[0]
for num in numbers:
    if num > max_num:
        max_num = num
print("Maximum:", max_num)

min_num = numbers[0]
for num in numbers:
    if num < min_num:
        min_num = num
print("Minimum:", min_num)

reversed_list = []
for i in range(len(numbers)-1, -1, -1):
    reversed_list.append(numbers[i])
print("Reversed List:", reversed_list)
```
