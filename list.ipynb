```python
"""Create two lists: a = [1, 2, 3], b = [4, 5, 6]
Perform:
Concatenation
Repetition (repeat list a 3 times)
Membership check (check if 2 is in list a)
Length of list"""

a=[1,2,3]
b=[4,5,6]
print(a+b)
print(a*3)
print(2 in a)
print(len(a))
print(len(b))
```

```python
""" nums = [10, 5, 20, 8, 15]
Perform:
Find maximum value
Find minimum value
Find sum of elements
Sort the list (ascending & descending)
Find index of element 8"""

nums=[10,5,20,8,15]
print(max(nums))
print(min(nums))
print(sum(nums))
print(sorted(nums))
descending=sorted(nums,reverse=True)
print(descending)

print(nums.index(8))
```

```python
""" Create a list: [10, 20, 30]
Perform:
Add element 40 using append()
Insert 15 at index 1
Remove element 20
Pop last element
Reverse the list"""

list=[10,20,30]
list.append(40)
print(list)
list.insert(1,15)
print(list)
list.remove(20)
print(list)
list.pop()
print(list)
list.reverse()
print(list)
```
