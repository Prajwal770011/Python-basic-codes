```python
#1. program to simulate a Train Seat Reservation System using a list.
def train_reservation_system():
    # Initialize a list of 20 seats with 0 (Available)
    seats = [0] * 20

    while True:
        print("\n--- Train Seat Reservation System ---")
        print("1. Book Seat")
        print("2. Cancel Seat")
        print("3. Show Seats")
        print("4. Booking Percentage")
        print("5. Exit")
        
        choice = input("Enter your choice (1-5): ")

        if choice == '1':
            # Book a seat
            seat_no = int(input("Enter seat number to book (1-20): ")) - 1
            if 0 <= seat_no < 20:
                if seats[seat_no] == 0:
                    seats[seat_no] = 1
                    print(f"Seat {seat_no + 1} booked successfully!")
                else:
                    print(f"Seat {seat_no + 1} is already booked.")
            else:
                print("Invalid seat number! Please choose between 1 and 20.")

        elif choice == '2':
            # Cancel a seat
            seat_no = int(input("Enter seat number to cancel (1-20): ")) - 1
            if 0 <= seat_no < 20:
                if seats[seat_no] == 1:
                    seats[seat_no] = 0
                    print(f"Seat {seat_no + 1} cancelled successfully!")
                else:
                    print(f"Seat {seat_no + 1} is already empty.")
            else:
                print("Invalid seat number!")

        elif choice == '3':
            # Display available seats
            print("\nCurrent Seat Map (0: Available, 1: Booked):")
            print(seats)
            available = [i + 1 for i, status in enumerate(seats) if status == 0]
            print(f"Available seat numbers: {available}")

        elif choice == '4':
            # Display booking percentage
            booked_count = seats.count(1)
            percentage = (booked_count / 20) * 100
            print(f"Total Booked: {booked_count}")
            print(f"Booking Percentage: {percentage}%")

        elif choice == '5':
            print("Thank you for using the system!")
            break
        
        else:
            print("Invalid choice! Please try again.")

if __name__ == "__main__":
    train_reservation_system()
```

```python
#2. YouTube Playlist Analysis using Lists

videos = [12, 4, 8, 15, 3, 20]

total_time = sum(videos)
longest_video = max(videos)
average_duration = total_time / len(videos)

short_videos = []

for video in videos:
    if video < 5:
        short_videos.append(video)

print("Video Durations:", videos)
print("Total Watch Time:", total_time, "minutes")
print("Longest Video:", longest_video, "minutes")
print("Average Video Duration:", average_duration, "minutes")
print("Videos shorter than 5 minutes:", short_videos)
```

```python
#3.Write a Python program for a Social Media Hashtag Tracker.
def track_hashtags(tags):
    frequency = {}
    for tag in tags:
        frequency[tag] = frequency.get(tag, 0) + 1
    
    most_repeated = max(frequency, key=frequency.get)
    unique_tags = list(set(tags))

    print(f"Original Tags: {tags}")
    print("-" * 30)
    print(f"Frequency of each tag: {frequency}")
    print(f"Most Repeated Tag:     {most_repeated} ({frequency[most_repeated]} times)")
    print(f"Unique Hashtags:       {unique_tags}")
    print(f"Total Unique Count:    {len(unique_tags)}")
sample_tags = ["#fun", "#python", "#fun", "#code", "#python"]
track_hashtags(sample_tags)
```

```python
#4. to perform Pattern Compression on a list.
def split_expenses(expenses):
    total_amount = sum(expenses)
    num_friends = len(expenses)
    average_share = total_amount / num_friends
    max_paid = max(expenses)
    highest_payer_index = expenses.index(max_paid) + 1 
    
    print("--- Expense Splitter Report ---")
    print(f"Individual Expenses:   {expenses}")
    print(f"Total Expenditure:     {total_amount}")
    print(f"Number of Friends:     {num_friends}")
    print("-" * 31)
    print(f"Highest Payment:       {max_paid} (Paid by Friend #{highest_payer_index})")
    print(f"Each Friend Owes:      {average_share:.2f}")

friend_expenses = [1200, 850, 1500, 950]
split_expenses(friend_expenses)
```

```python
#5. program to perform Pattern Compression on a list.
def compress_pattern(data):
    if not data:
        return []
    compressed = []
    count = 1
    for i in range(1, len(data)):
        if data[i] == data[i - 1]:
            count += 1
        else:
            compressed.append((data[i - 1], count))
            count = 1  
    compressed.append((data[-1], count))
    
    return compressed
    
input_list = [1, 1, 1, 2, 2, 3, 3, 3,4]
output = compress_pattern(input_list)

print(f"Input : {input_list}")
print(f"Output: {output}")
```

```python
###### 6. Develop a Mini Shopping Cart System using lists 

def shopping_cart():
    cart = []
    
    while True:
        print("\n--- Mini Shopping Cart ---")
        print("1. Add Item")
        print("2. Remove Item")
        print("3. Search Item")
        print("4. Calculate Total Bill")
        print("5. Display Duplicate Items")
        print("6. Exit")
        
        choice = input("Select an option: ")

        if choice == '1':
            item = input("Enter item name: ").strip().lower()
            price = float(input(f"Enter price for {item}: "))
           
            cart.append((item, price))
            print(f"'{item}' added to cart.")

        elif choice == '2':
            item_to_remove = input("Enter item name to remove: ").strip().lower()
        
            initial_count = len(cart)
            cart = [item for item in cart if item[0] != item_to_remove]
            
            if len(cart) < initial_count:
                print(f"All instances of '{item_to_remove}' removed.")
            else:
                print("Item not found in cart.")

        elif choice == '3':
            search_item = input("Search for: ").strip().lower()
            
            names = [item[0] for item in cart]
            if search_item in names:
                count = names.count(search_item)
                print(f"'{search_item}' is in your cart ({count} units).")
            else:
                print("Item not found.")

        elif choice == '4':
            
            total = sum(item[1] for item in cart)
            print(f"Current Total Bill: ${total:.2f}")

        elif choice == '5':
        
            names = [item[0] for item in cart]
            duplicates = set([name for name in names if names.count(name) > 1])
            
            if duplicates:
                print(f"Items added more than once: {', '.join(duplicates)}")
            else:
                print("No duplicate items found.")

        elif choice == '6':
            print("Exiting... Happy Shopping!")
            break
        else:
            print("Invalid selection. Try again.")

if __name__ == "__main__":
    shopping_cart()
```
