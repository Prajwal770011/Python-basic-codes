```python
#LIBRARY MANAGEMENT SYSTEM
""" Problem statement 1
  create a LIBRARY MANAGEMENT SYSTEM using OOP concept to store and display book details"""
class book:
    def __init__(bobo,title,aurthor,price):
        bobo.title = title
        bobo.aurthor = aurthor
        bobo.price = price
    def display(bobo):
        print("Title",bobo.title)
        print("Aurthor",bobo.aurthor)
        print("Price",bobo.price)
s1 = book("THE ONE PIECE","LUFFY",500000)
s2 = book("BLACK CLOVER","YAMI",5156)
s1.display()
print("________------('')-------___________")
s2.display()
```

```python
""" Problem statement 2
    create a bank system with deposit and withdraw operation"""

class bank:
    def __init__(aaa,name,balence):
        aaa.name = name
        aaa.balence = balence

    def deposit(aaa,amount):
         aaa.balence <+ amount
         print(amount,"deposited successfully")

    def withdraw(aaa,amount):
         if amount <= aaa.balence:
             aaa.balence -= amount
             print(amount,"withdrawn successfully")
         else:
               print("Insufficient balance")
             
    def display(aaa):
        print("Account holder:",aaa.name)
        print("Balance",aaa.balence)

b = bank("Prajwal",654156132689168)
b.display()
b.deposit(6511) 
b.withdraw(5131)
        
```

```python
""" 3 Student result management system"""

class student:
    def __init__(lala,name,m1,m2,m3):
        lala.name = name
        lala.m1 = m1
        lala.m2 = m2
        lala.m3 = m3
    def calcuate(lala):
        total = lala.m1+lala.m2+lala.m3
        pres = total / 3

        print("Student Name:",lala.name)
        print("Total marks   :",total)
        print("Percentage    :",pres)
        
s1 = student("Prajwal",100,99,100)
s1.calcuate()
```
