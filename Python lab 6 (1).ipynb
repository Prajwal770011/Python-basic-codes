```python
                               #FUNCTION

# 1. simple function
def greet():
    print("Hello Word!")
greet()

#2 .function with parameters
def add(a,b):
    return a+b
print(add(3,5))

# 3.Function with default argument
def power(base,exp=2):
    return base**exp
print(power(4))
print(power(4,3))

#4.Function returning multiple values
def stats (numbers):
    return min(numbers),max(numbers),sum(numbers) / len(numbers)
print(stats([2,4,6,8]))

#5 .lambda functions
square = lambda x: x * x
print(square(654))  
```

```python
                             #CALLBACK

#1 .Basic callback
def process(data,callback):
    result=data*2
    callback(result)

def showoutput(value):
    print("Processed:",value)

process(5,showoutput)

#2 .callback with lambda
process(10,lambda x: print("Lambda Output:", x))
```

```python
                              #RECURSION
#1 simple function
def hello(n):
    if n > 0:
        print("Hello World")
        hello(n-1)

hello(3)
```
